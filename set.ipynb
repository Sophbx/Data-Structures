# Set (Python3)
## Stores unique, hashable elements 
Quickly check whether an element is present.

```python
# Initiate a set
seen = set()              # Empty set. {} creates an empty dictionary.
numbers = {1, 2, 3}        # Set with three elements
unique = set([1, 1, 2])    # {1, 2}

print(seen)
print(numbers)
print(unique)
```

```python
# Add elements to a set (if element x is absent)
seen.add(4)
print(seen)
seen.add(4)                # Adding it again leaves the set unchanged
print(seen)
```

```python
# Membership test
print(4 in seen)           # True
```

```python
# Size
print(len(seen))           # 1
```

```python
# Check if empty
seen = []
if not seen:
    print("Set is empty")
bool(seen)
```

```python
# Safely / Required remove
numbers = {1, 2, 3} 

numbers.discard(1)  # Remove 1 from the set.
numbers.discard(5)  # 5 is not present, do nothing.
print(numbers)  # Output: {2, 3}

numbers.remove(2)  # Remove 2 from the set.
print(numbers)
try:
    numbers.remove(4)  # 4 is not present, raise KeyError.
except KeyError:
    print("Element not found in the set.")
```
