# Dictionary (Python3)
## key-value pairs.
use a key to look up its associated value.

```python
# Define
empty_dict = {}  # An empty dictionary

ages = {"Alice": 20,
        "Bob": 22}
```

`key` must be unique, `value` can repaet.

```python
# Access values using keys
print(ages["Alice"])  
print(ages.get("Charlie"))     # Returns None
print(ages.get("Charlie", 0))  # Returns 0 if "Charlie" is missing

print(ages)
```

```python
# Add / Update
ages["Charlie"] = 21  # Adds a new key–value pair
ages["Alice"] = 25    # Replaces Alice's existing value

print(ages)
```

```python
# Check if a key exists
print("Alice" in ages)  # True
print("David" in ages)  # False
```

```python
# Check if a value exists
print(25 in ages.values())  # True
```

If each key needs two values, store a **container as value**. The dictionary works `one key : one value`, but that value object can hold multiple pieces of information.

```python
# e.g. Tuple
students = {"Alice": (20, 95),
            "Bob": (22, 88)}

print(students["Alice"])     # (20, 95)
print(students["Alice"][0])  # 20
print(students["Alice"][1])  # 95

age, score = students["Alice"]  # Unpacks the pair into two variables

students["Alice"] = (21, 98)    # Tuples cannot be modified in place, can replace the entire pair
print(students["Alice"])     
```

```python
# e.g.: List
# Lists allows in-place modification of values
students = {"Alice": [20, 95]}

students["Alice"][1] = 98
print(students["Alice"])  # [20, 98]
```

```python
# e.g.: Dictionary -- when descriptive name desired
students = {"Alice": {"age": 20, "score": 95}}

print(students["Alice"]["score"])  # 95
students["Alice"]["score"] = 98
print(students["Alice"]["score"]) 
```

```python
# Loop through a dictionary
ages = {"Alice": 20, "Bob": 22}

for x in ages:
    print(x)              # Each key

for x in ages.values():
    print(x)               # Each value

for name, age in ages.items():
    print(name, age)         # Each key and its associated value
```

```python
# lenth and remove
print(len(ages))         # 2
del ages["Bob"]
print(len(ages))         # 1
```
