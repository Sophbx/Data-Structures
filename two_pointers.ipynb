# Two Pointers (Python3)
## Simply integer indices.
Treat them as offset.

```python
# Define
text = "level"

left = 0
right = len(text) - 1
```

```python
# Very simple example with palindrome check
while left < right:
    if text[left] != text[right]:
        print("Not a palindrome")
        break
    left += 1
    right -= 1

print("Is a palindrome")
```
