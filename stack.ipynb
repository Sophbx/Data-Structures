# Stack (Python3)
## Last In, First Out (LIFO).
In Python, use a regular list. The end of the list is the top of the stack.

```python
# Initiate an empty stack
stack = []
```

```python
# Push
stack.append("a")  # Push "a"
stack.append("b")  # Push "b"
stack.append("c")  # Push "c"
print(stack)
```

```python
# Peek
stack[-1]          # "c" — peek at the top
print(stack[-1])   
```

```python
# Return / remove
print(stack.pop())        # "c" — remove and return the top
print(stack.pop())        # "b"
print(stack)
```

```python
# Check if empty
if not stack:
    print("Stack is empty")

stack = []
print(bool(stack))
```

Emptiness is different from reference/type

```python
print(stack is stack)
print(stack is not stack)
```

```python
# Get the size of the stack
stack = ["a", "b", "c"]
print(len(stack))  # Output: 3
```
