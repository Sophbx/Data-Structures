# Priority Queue (Python3) with heapq
## Retrieves items by priority.
Python’s heapq operates on a regular list and gives the smallest priority first.     
Use (priority, item) tuples to store both pieces of information.

```python
import heapq
```

```python
# Initialize a regular list that we will run heapq on.
tasks = []
```

heappush() appends the new item, then moves it up by swapping it with its `parent` whenever its priority is smaller. 

```python
# Add: heapq.heappush(heap, item)
heapq.heappush(tasks, (3, "write report"))
print(tasks)  # [(3, "write report")]

heapq.heappush(tasks, (1, "fix bug"))
heapq.heappush(tasks, (2, "review code"))
print(tasks)  # [(1, "fix bug"), (3, "write report"), (2, "review code")]
```

For an item at `index i > 0`, its `parent` is at `(i - 1) // 2`. So when 2 lands at index 2, its parent is 1 at index 0. The 3 is its `sibling`, and **siblings can appear in either order**.

```python
# Get/access
tasks[0]                 # (1, "fix bug"), without removing it --> Peak at the smallest item
print(tasks[0])  
```

```python
# Return/remove: heapq.heappop(heap) --> Removes and returns the smallest item
print(heapq.heappop(tasks))      # (1, "fix bug"), removes it
print(tasks)

print(heapq.heappop(tasks))      # (2, "review code"), removes it
print(tasks)
```

```python
# Turns an existing list into a heap in place
list = [5, 7, 9, 1, 3]
heapq.heapify(list)
print(list) # [1, 3, 9, 7, 5] --> The smallest item is at index 0
```
