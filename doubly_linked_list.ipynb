# Doubly Linked List (Python3)
## Nodes with prev and next links
If you already have a reference to a node, you can use those two links to remove it in O(1) time.


```python
# To define/initiate a node
class Node:
    def __init__(self, val):
        self.val = val
        self.prev = None
        self.next = None
```

```python
# To define/initiate a doubly linked list
class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
```

```python
class DoublyLinkedList:
    def append(self, node):
        """Add a node to the end of the list."""
        # node should be detached from any list.
        node.prev = self.tail

        if self.tail is None:
            self.head = node
        else:
            self.tail.next = node

        self.tail = node

    def remove(self, node):
        """Remove a node from the list."""
        # node should already belong to this list.
        previous = node.prev
        following = node.next

        if previous is None:
            self.head = following
        else:
            previous.next = following

        if following is None:
            self.tail = previous
        else:
            following.prev = previous

        # Clean the node refernce and links as well
        node.prev = None
        node.next = None
```
