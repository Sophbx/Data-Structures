# OrderedDict (Python3)
## Stores key–value pairs and move a key to either end

```python
from collections import OrderedDict
```

```python
# Create an empty one
test = OrderedDict()

# Create by pass in a list
# Each element in the list is a tuple of key-value
items = OrderedDict([("a", 10), ("b", 20), ("c", 30)])
```

```python
# Get/access
print(items["a"])                     # 10
print(items.get("missing"))           # No such key, None
print(items.get("missing", -1))       # -1 if None
print("b" in items)                   # True
```

```python
# Update
items["a"] = 11                # Updates the value. Order is still a, b, c.
```

```python
# Order: move_to_end(key, last=True/False)
# Moves an existing key to the back
items.move_to_end("a")         # Order is now b, c, a.
print(items)

# Move an existing key to the front
items.move_to_end("a", last=False)  # Order is now a, b, c.
print(items)
```

```python
# Return/remove
# Removes and returns the last (key, value) pair
print(items.popitem())
items = OrderedDict([("b", 20), ("c", 30), ("a", 10)])

print(items.popitem(last=False))      # Returns first element ("b", 20) and removes it.
```
