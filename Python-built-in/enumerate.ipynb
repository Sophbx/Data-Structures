# Enumerate
## Provide a counter and the item itself as you loop. 
The counter starts at 0 by default, so for a list it matches the index.

```python
# Example
seats = [1, 0, 0, 1]

for index, seat in enumerate(seats):
    print(index, seat)
```

On each iteration, enumerate() yields a pair like (1, 0). Python unpacks it into index = 1 and seat = 0.
