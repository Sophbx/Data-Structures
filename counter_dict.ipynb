# Counter
## Maps each item to its frequency.
A dictionary from Python’s collections module.     
Looking up a missing key with `[]` gives 0, so you can increment a count without initializing it first.

```python
from collections import Counter
```

```python
# Initiate
chars = Counter()        # Start with no counts
chars["a"] += 1          # 1
chars["a"] += 1          # 2
print(f"b: {chars['b']}")               # 0
print(chars)
```

```python
Counter("banana")        # Counter({'a': 3, 'n': 2, 'b': 1})
```
